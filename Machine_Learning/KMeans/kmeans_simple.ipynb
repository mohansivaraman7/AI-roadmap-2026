```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd

# Load iris dataset
iris = load_iris()
X = iris.data  # We only use the features, not the target
y = iris.target  # For comparison only

# Create KMeans model with 3 clusters (since iris has 3 species)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

# Predict clusters
labels = kmeans.labels_

# Compare with actual labels (just for validation, not part of clustering)
print("Cluster labels assigned by KMeans:")
print(labels)

# Visualize using first two features
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='magma', marker='o')
plt.title("KMeans Clustering on Iris Dataset (first two features)")
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.colorbar()
plt.grid(True)
plt.show()
```
